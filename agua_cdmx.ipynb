## Problema y contexto

La pregunta de investigación que guía este cuaderno es la siguiente: **¿es posible predecir el gasto total de agua por colonia usando únicamente el número de hogares por colonia?** La pregunta no es trivial. En 2024, la Ciudad de México vivió uno de los momentos más delicados de su discusión pública sobre disponibilidad de agua: niveles críticamente bajos en el Sistema Cutzamala, temores alrededor del llamado *día cero*, tandeos, presión sobre la infraestructura urbana y una conversación cada vez más visible sobre desigualdad en el acceso, fugas y sostenibilidad del modelo hídrico metropolitano. La relevancia pública del tema quedó reflejada en medios de amplia circulación nacional e internacional, por ejemplo en BBC, CNN en Español y *El País*, que documentaron tanto el riesgo de desabasto como la complejidad estructural del problema.

## Origen y propósito de los datos

Los datos utilizados en este análisis provienen del portal de **Datos Abiertos de la Ciudad de México** (<https://datos.cdmx.gob.mx/>), una plataforma oficial que concentra información pública generada por distintas dependencias del gobierno capitalino. En particular, este cuaderno trabaja con un conjunto de datos de **consumo histórico de agua** y con una capa territorial de **hogares por colonia**, utilizada para homologar el consumo con una medida demográfica básica. El propósito de estas fuentes es permitir una lectura territorial del comportamiento del consumo de agua y, al mismo tiempo, construir una base analítica que haga posible evaluar si el número de hogares por colonia funciona como predictor del consumo total. En términos metodológicos, ambas fuentes se integran para producir una unidad de análisis comparable a nivel colonia y semestre, adecuada para limpieza, exploración y modelado simple.

## Necesidad o problemática a resolver

Predecir el consumo de agua por colonia puede aportar valor práctico en varios frentes. Primero, permite construir una línea base cuantitativa para entender cómo se distribuye la demanda en el territorio urbano. Segundo, ayuda a identificar si una variable de fácil acceso —el número de hogares— captura por sí sola una parte significativa del patrón de consumo o si, por el contrario, resulta claramente insuficiente. Esta distinción importa porque, en contextos de crisis hídrica, los gobiernos, operadores y equipos técnicos necesitan herramientas simples, rápidas y replicables para estimar demanda, priorizar zonas de atención y detectar comportamientos anómalos. Sin embargo, también es importante no sobreprometer: si el número de hogares explica poco, entonces la planeación hídrica requiere incorporar otras dimensiones como mezcla de usos residencial/comercial, densidad construida, ingreso, fugas, infraestructura y patrones diferenciados de consumo. En ese sentido, el problema no consiste sólo en “hacer una regresión”, sino en evaluar con rigor hasta dónde llega el poder explicativo de una variable única y cuándo deja de ser suficiente para describir una ciudad tan heterogénea como la CDMX. Este notebook busca justamente responder eso con una base depurada y criterios explícitos de limpieza, exclusión de outliers y comparación de modelos.

## Importancia del análisis

La importancia de este análisis es doble. Por un lado, tiene valor sustantivo: en una ciudad que enfrentó una crisis hídrica visible en 2024, estimar y entender patrones de consumo no es un ejercicio abstracto, sino una necesidad ligada a gestión pública, planeación y resiliencia urbana. Por otro lado, tiene valor metodológico: permite poner a prueba una hipótesis parsimoniosa y verificar si una sola variable demográfica puede funcionar como predictor base del consumo total. Incluso si el resultado muestra que el ajuste es limitado, ese hallazgo sigue siendo útil, porque ayuda a delimitar qué tan lejos se puede llegar con modelos simples y qué variables adicionales serían necesarias en una etapa posterior.

```python
# Importamos librerías para manipulación, geodatos, visualización y apoyo al matching difuso.
import pandas as pd
import geopandas as gpd
import unicodedata
import re
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from difflib import SequenceMatcher

```

```python
# Definimos rutas de entrada y salida dentro de la carpeta del proyecto.
base_path = Path('.')
agua_path = base_path / 'consumo_agua' / 'consumo_agua_historico_2019.csv'
hogares_path = base_path / 'Hogares por colonia' / 'hogares_colonia.shp'
output_bimestre_path = base_path / 'consumo_agua' / 'consumo_agua_por_colonia_bimestre_con_hogares_2020.csv'
output_semestre_path = base_path / 'consumo_agua' / 'consumo_agua_por_colonia_semestre_con_hogares_2020.csv'
output_semestre_filtrado_path = base_path / 'consumo_agua' / 'consumo_agua_por_colonia_semestre_con_hogares_2020_filtrado.csv'
```

```python
# Función para normalizar textos y facilitar cruces por nombre de colonia/alcaldía.
def normalizar_texto(valor):
    if pd.isna(valor):
        return pd.NA
    texto = str(valor).strip().upper()
    texto = unicodedata.normalize('NFKD', texto).encode('ASCII', 'ignore').decode('ASCII')
    texto = re.sub(r'\s+', ' ', texto)
    return texto

# Función auxiliar para detectar firmas de sección (I, II, 1ERA, etc.)
# y evitar matches difusos peligrosos entre colonias parecidas.
def extraer_firma_seccion(texto):
    if pd.isna(texto):
        return tuple()
    patrones = re.findall(r'(?:[IVXLCDM]+|\d+(?:ERA|DA|RA|TA|MA|A)?(?:-\d+)?)', str(texto))
    return tuple(patrones)
```

### Estructura y variables del dataset de agua

El dataset de consumo de agua contiene registros administrativos con información temporal, territorial y de consumo. En esta etapa interesa describir su **estructura original completa**, antes de cualquier depuración, para dejar documentadas **todas** las variables disponibles en la fuente.

En otras palabras, aquí se documenta el dataset **tal como viene en origen**. Más adelante se indicará cuáles variables se conservan para el flujo analítico y cuáles se descartan por no ser necesarias en las etapas posteriores.

La siguiente tabla resume nombre, tipo, rango observado y valores faltantes para todas las variables del dataset de agua al momento de la carga.


```python
# Tabla descriptiva de variables del dataset de agua.
def resumir_variables(df):
    filas = []
    for col in df.columns:
        serie = df[col]
        tipo = str(serie.dtype)
        faltantes = int(serie.isna().sum())
        if pd.api.types.is_numeric_dtype(serie):
            rango = f"{serie.min(skipna=True)} a {serie.max(skipna=True)}"
        else:
            no_na = serie.dropna().astype(str)
            if no_na.empty:
                rango = 'sin valores observados'
            else:
                ejemplos = no_na.unique()[:5]
                rango = ' | '.join(ejemplos)
        filas.append({
            'variable': col,
            'tipo': tipo,
            'rango_o_valores': rango,
            'valores_faltantes': faltantes
        })
    return pd.DataFrame(filas)

resumen_variables_agua = resumir_variables(df_agua)
resumen_variables_agua
```

### Depuración temprana de columnas

Antes de seguir, recortamos el dataset original para quedarnos sólo con las variables que sí alimentan el resto del flujo.

Se eliminan columnas como promedios de consumo e `indice_des` porque no se vuelven a usar en agregaciones, cruces o visualizaciones posteriores.

**Nota:** `fecha_referencia`, `anio`, `bimestre`, `consumo_total_mixto`, `consumo_total_dom` y `consumo_total_no_dom` se conservan en esta etapa porque todavía participan en la agregación bimestral y en la construcción de los datasets de salida.


```python
# Eliminamos columnas del archivo original que no se usan más adelante.
columnas_descartables = [
    'consumo_prom_dom',
    'consumo_prom_mixto',
    'consumo_prom',
    'consumo_prom_no_dom',
    'indice_des'
]

columnas_descartadas = [col for col in columnas_descartables if col in df_agua.columns]
df_agua = df_agua.drop(columns=columnas_descartadas).copy()

print('Columnas eliminadas:', columnas_descartadas)
print('Columnas restantes:', df_agua.columns.tolist())
```

```python
# Eliminamos duplicados exactos porque no aportan información nueva y distorsionan agregaciones.
duplicados_exactos = df_agua.duplicated().sum()
df_agua = df_agua.drop_duplicates().copy()

print(f'Duplicados exactos eliminados: {duplicados_exactos:,}')
print(f'Filas después de limpiar: {len(df_agua):,}')
```

```python
# Agregamos el consumo a nivel colonia-bimestre para quitar la granularidad interna
# y trabajar con una unidad territorial consistente para el cruce posterior.
columnas_agrupacion = ['fecha_referencia', 'anio', 'bimestre', 'alcaldia', 'colonia']
columnas_suma = [
    'consumo_total_mixto',
    'consumo_total_dom',
    'consumo_total',
    'consumo_total_no_dom'
]

df_agua_colonia = (
    df_agua
    .groupby(columnas_agrupacion, as_index=False)[columnas_suma]
    .sum()
)

# Generamos versiones normalizadas de alcaldía y colonia para hacer merges más robustos.
df_agua_colonia['alcaldia_norm'] = df_agua_colonia['alcaldia'].apply(normalizar_texto)
df_agua_colonia['colonia_norm'] = df_agua_colonia['colonia'].apply(normalizar_texto)

print(f'Filas agregadas a nivel colonia-bimestre: {len(df_agua_colonia):,}')
df_agua_colonia.head()
```

### Transformaciones de datos aplicadas

Antes de continuar con el cruce y la construcción de la base analítica, conviene documentar tres transformaciones centrales aplicadas a los datos.

**1. Normalización de nombres territoriales.** Se generaron versiones normalizadas de `alcaldia` y `colonia` para estandarizar el texto, convirtiéndolo a mayúsculas, eliminando acentos y reduciendo espacios redundantes. Esta transformación permite homologar nombres entre fuentes distintas y disminuir errores en los cruces por diferencias ortográficas o de captura.

**2. Integración de fuentes de datos.** El análisis combina dos fuentes: el dataset de consumo de agua y el shapefile de hogares por colonia. Para ello, se integran ambas bases a partir de identificadores territoriales, apoyándose en nombres normalizados y, cuando fue necesario, en procedimientos de matching difuso. Esta transformación permite construir una sola base analítica con información de consumo y número de hogares por colonia.

**3. Recodificación del total de hogares.** La variable `Sum_TotHog` se convirtió a formato numérico entero nullable (`Int64`), aplicando coerción a valores no válidos y redondeo cuando fue necesario. ## 3. Cruce exacto y matching difuso de colonias

En esta etapa se vinculan las dos fuentes principales del análisis: la base de consumo de agua agregada por colonia y la base de hogares por colonia. El objetivo es asignar correctamente a cada registro de consumo una clave territorial (`cve_col`) y un total de hogares (`Sum_TotHog`) para poder construir una unidad analítica consistente.

El procedimiento se realiza en dos niveles. Primero, se intenta un **cruce exacto** usando los nombres normalizados de `alcaldia` y `colonia`. Esta vía resuelve la mayoría de los casos cuando ambas fuentes nombran las colonias de manera suficientemente similar. Después, para los registros que siguen sin correspondencia, se aplica un **matching difuso** dentro de la misma alcaldía. Este segundo paso compara similitud textual entre nombres de colonias y ayuda a recuperar coincidencias que no aparecen de forma idéntica por abreviaturas, errores de captura, variantes ortográficas o diferencias de acentuación.

Además, el proceso incluye una revisión para evitar falsos positivos entre colonias con nombres parecidos pero distintas secciones o denominaciones. Por eso, el matching difuso no se usa como sustituto automática del cruce exacto, sino como una capa complementaria y controlada. Esta etapa es clave porque de su calidad depende que el consumo quede correctamente asociado con el número de hogares y que el análisis posterior no herede errores de homologación territorial.


```python
# Hacemos primero el cruce exacto usando alcaldía y colonia normalizadas.
df_match_exacto = df_agua_colonia.merge(
    df_hogares[['alcaldia_norm', 'colonia_norm', 'cve_col', 'Sum_TotHog']],
    on=['alcaldia_norm', 'colonia_norm'],
    how='left'
)

print(f"Filas con match exacto o por normalización: {df_match_exacto['Sum_TotHog'].notna().sum():,}")
print(f"Filas sin match tras cruce exacto: {df_match_exacto['Sum_TotHog'].isna().sum():,}")
df_match_exacto.head()
```

### Paso 1: identificar colonias que no resolvieron con el cruce exacto

Después del merge exacto, todavía quedan colonias del dataset de agua que no encuentran correspondencia en la base de hogares. Esto puede ocurrir por diferencias mínimas en escritura, abreviaturas, errores tipográficos, variantes con o sin acentos, o porque una colonia aparece con una denominación ligeramente distinta entre ambas fuentes.

Para no perder esos registros, primero se aísla la lista de colonias sin match exacto. La idea es trabajar sólo sobre los casos problemáticos y no volver a tocar los cruces que ya quedaron resueltos correctamente. Este paso también reduce el riesgo de introducir ruido innecesario, porque el matching difuso se reserva para un subconjunto pequeño y controlado del universo total.


```python
# Aislamos únicamente las colonias que siguen sin match exacto.
# Este subconjunto será la entrada del procedimiento de fuzzy matching.
colonias_sin_match = (
    df_match_exacto.loc[df_match_exacto['Sum_TotHog'].isna(), ['alcaldia', 'colonia', 'alcaldia_norm', 'colonia_norm']]
    .drop_duplicates()
    .sort_values(['alcaldia', 'colonia'])
    .reset_index(drop=True)
)

print(f'Colonias sin match de hogares antes de fuzzy: {len(colonias_sin_match):,}')
colonias_sin_match[['alcaldia', 'colonia']]
```

### Paso 2: generación de candidatos con fuzzy matching

En esta celda se construye el corazón del procedimiento de matching difuso. Para cada colonia sin correspondencia exacta, se buscan candidatos sólo dentro de la **misma alcaldía**, lo cual reduce drásticamente el espacio de comparación y evita empates absurdos con colonias de otras demarcaciones.

Luego se calcula un `score` de similitud textual con `SequenceMatcher`, que produce un valor entre 0 y 1: mientras más cercano a 1, más parecidos son los nombres. Además, no sólo se toma el mejor candidato; también se guarda el segundo mejor. Esto es importante porque permite medir el `gap` entre ambos y distinguir entre coincidencias claras y coincidencias ambiguas.

Por último, se agrega una validación específica sobre la **firma de sección** (`I`, `II`, `1ERA`, etc.). Esta regla evita falsos positivos en colonias cuyos nombres base son casi iguales, pero que pertenecen a secciones distintas. El resultado final de la celda es una tabla ordenada con el mejor candidato por colonia, sus puntajes y una bandera `auto_match` que marca sólo los casos suficientemente confiables para aceptarse automáticamente.


```python
# Para cada colonia sin match exacto, calculamos similitud contra colonias de la misma alcaldía.
# También guardamos la firma de sección para evitar falsos positivos entre colonias parecidas.

candidatos_fuzzy = []

for _, fila in colonias_sin_match.iterrows():
    # Reducimos el espacio de búsqueda a la misma alcaldía para comparar sólo candidatos plausibles.
    candidatos_alcaldia = df_hogares.loc[
        df_hogares['alcaldia_norm'] == fila['alcaldia_norm'],
        ['cve_col', 'colonia', 'colonia_norm']
    ].drop_duplicates()

    comparaciones = []
    firma_origen = extraer_firma_seccion(fila['colonia_norm'])

    for _, candidato in candidatos_alcaldia.iterrows():
        # Score de similitud textual entre el nombre de colonia en agua y el candidato en hogares.
        score = SequenceMatcher(None, fila['colonia_norm'], candidato['colonia_norm']).ratio()
        firma_candidato = extraer_firma_seccion(candidato['colonia_norm'])
        comparaciones.append({
            'score': score,
            'cve_col_match': candidato['cve_col'],
            'colonia_match': candidato['colonia'],
            'firma_origen': firma_origen,
            'firma_candidato': firma_candidato,
            'firma_compatible': firma_origen == firma_candidato
        })

    # Ordenamos de mayor a menor similitud para recuperar el mejor y el segundo mejor candidato.
    comparaciones = sorted(comparaciones, key=lambda x: x['score'], reverse=True)
    mejor = comparaciones[0]
    segundo = comparaciones[1] if len(comparaciones) > 1 else None
    score_segundo = segundo['score'] if segundo else 0

    candidatos_fuzzy.append({
        'alcaldia': fila['alcaldia'],
        'colonia_agua': fila['colonia'],
        'score_top1': round(mejor['score'], 4),
        'colonia_top1': mejor['colonia_match'],
        'cve_col_top1': mejor['cve_col_match'],
        'score_top2': round(score_segundo, 4),
        'colonia_top2': segundo['colonia_match'] if segundo else pd.NA,
        'cve_col_top2': segundo['cve_col_match'] if segundo else pd.NA,
        'gap_top12': round(mejor['score'] - score_segundo, 4),
        'firma_compatible': mejor['firma_compatible'],
        # Sólo aceptamos como match automático casos muy fuertes:
        # score alto, firma compatible y diferencia suficiente respecto al segundo mejor candidato.
        'auto_match': (mejor['score'] >= 0.97) and mejor['firma_compatible'] and ((mejor['score'] - score_segundo) >= 0.02)
    })

df_fuzzy = pd.DataFrame(candidatos_fuzzy).sort_values(['auto_match', 'score_top1', 'gap_top12'], ascending=[False, False, False]).reset_index(drop=True)

print(f"Matches automáticos por fuzzy: {df_fuzzy['auto_match'].sum():,}")
print(f"Casos para revisión manual: {(~df_fuzzy['auto_match']).sum():,}")
df_fuzzy.head(20)
```

```python
# Extraemos sólo los casos que el criterio considera suficientemente confiables como auto_match.
# Después les añadimos Sum_TotHog a partir de la clave territorial encontrada.
matches_fuzzy_automaticos = df_fuzzy.loc[
    df_fuzzy['auto_match'],
    ['alcaldia', 'colonia_agua', 'cve_col_top1']
].rename(columns={'colonia_agua': 'colonia', 'cve_col_top1': 'cve_col'})

matches_fuzzy_automaticos = matches_fuzzy_automaticos.merge(
    df_hogares[['cve_col', 'Sum_TotHog']].drop_duplicates(),
    on='cve_col',
    how='left'
)

matches_fuzzy_automaticos
```

### Paso 3: separación entre matches automáticos y revisión manual

Una vez calculados los puntajes de similitud, el flujo separa dos tipos de resultados. Por un lado están los **matches automáticos**, es decir, casos con similitud muy alta, firma compatible y una diferencia suficiente frente al segundo mejor candidato. ## 7. Identificación y exclusión de casos extremos para el modelado con consumo total

Aquí se construye una capa de depuración para separar colonias que pueden distorsionar el análisis. En esta etapa se identifican los casos extremos que se excluirán de la base limpia usada en el EDA y en el modelado posterior.

### Qué se excluye

Se retiran del dataset limpio tres tipos de casos:

1. colonias con `Sum_TotHog <= 0`,
2. colonias con consumo total por hogar por encima del percentil 99,
3. colonias agregadas manualmente por comportamiento atípico claro.

### Nota metodológica

Las exclusiones manuales no significan necesariamente que el dato sea "falso". En varios casos, el problema es analítico: hay colonias con tamaños extremos o configuraciones muy particulares que aplastan la nube de puntos y vuelven menos interpretable la relación general entre hogares y consumo.

Por eso conviene separar:

- **base completa** para trazabilidad,
- **base filtrada** para EDA y modelado.


### Criterios y lógica de identificación de casos extremos

Este bloque define qué colonias se excluirán de la base limpia antes del modelado y del análisis exploratorio final. La lógica combina **criterios estadísticos** y **criterios sustantivos**.

Primero, se calcula un umbral de referencia usando el **percentil 99** de `consumo_total_por_hogar_sem`. Esto permite identificar colonias cuyo consumo por hogar es extraordinariamente alto respecto del resto de la distribución, sin depender únicamente de inspección visual.

Segundo, se incorpora una lista manual de colonias con **tamaños extremos de hogares**. Estos casos no necesariamente son los más altos en consumo por hogar, pero sí deforman la escala de gráficas y relaciones estadísticas, especialmente cuando concentran volúmenes atípicos de hogares. Por eso se marcan explícitamente y no se dejan al azar de una sola regla estadística.

Después se construye una máscara booleana que reúne tres condiciones: colonias con `Sum_TotHog <= 0`, colonias por encima del percentil 99 en consumo por hogar y colonias incluidas en la lista manual de outliers por hogares extremos. Finalmente, esa máscara se aplica sobre `df_semestre_stats` para extraer una tabla ordenada con las colonias excluidas. El propósito no es borrar información, sino separar una base más estable para el EDA y el modelado, manteniendo trazabilidad sobre qué casos fueron retirados y por qué.


```python
# Paso 1: calculamos el percentil 99 del consumo total semestral por hogar.
# Este umbral sirve como regla estadística para detectar colonias con valores
# extraordinariamente altos frente al resto de la distribución.
umbral_p99 = df_semestre_stats['consumo_total_por_hogar_sem'].quantile(0.99)

# Paso 2: definimos una lista manual de colonias con tamaños extremos de hogares.
# Estos casos se agregan explícitamente porque deforman la escala de las gráficas
# o introducen comportamientos territoriales poco comparables con el resto.
outliers_hogares_extremos = [
    {'alcaldia': 'IZTAPALAPA', 'colonia': 'LOMAS DE SAN LORENZO II'},
    {'alcaldia': 'CUAUHTEMOC', 'colonia': 'CENTRO VIII'},
    {'alcaldia': 'GUSTAVO A. MADERO', 'colonia': 'LOMA DE LA PALMA'},
    {'alcaldia': 'IZTAPALAPA', 'colonia': 'SANTA MARTHA ACATITLA_(PBLO)'},
    {'alcaldia': 'XOCHIMILCO', 'colonia': 'SAN MATEO XALPA (PBLO)'}
]
df_outliers_hogares_extremos = pd.DataFrame(outliers_hogares_extremos)

# Paso 3: construimos una máscara booleana para marcar esas colonias manuales.
# La serie empieza en False para todas las filas y se actualiza a True cuando
# una observación coincide en alcaldía y colonia con algún caso de la lista.
mask_outliers_hogares_extremos = pd.Series(False, index=df_semestre_stats.index)
for _, fila in df_outliers_hogares_extremos.iterrows():
    mask_outliers_hogares_extremos |= (
        (df_semestre_stats['alcaldia'] == fila['alcaldia'])
        & (df_semestre_stats['colonia'] == fila['colonia'])
    )

# Paso 4: combinamos todos los criterios de exclusión en una sola máscara final.
# Una colonia se marca como caso extremo si:
# - no tiene hogares válidos,
# - supera el percentil 99 en consumo total por hogar,
# - o pertenece a la lista manual de hogares extremos.
mask_casos_extremos = (
    (df_semestre_stats['Sum_TotHog'] <= 0)
    | (df_semestre_stats['consumo_total_por_hogar_sem'] > umbral_p99)
    | mask_outliers_hogares_extremos
)

# Paso 5: extraemos en una tabla sólo las colonias que cumplen al menos uno de
# los criterios anteriores. Esta tabla funciona como registro explícito de qué
# casos se excluyen y por qué se separan de la base limpia.
casos_extremos_excluidos = df_semestre_stats.loc[mask_casos_extremos, [
    'alcaldia', 'colonia', 'Sum_TotHog', 'consumo_total', 'consumo_total_por_hogar_sem'
]].sort_values(['alcaldia', 'colonia']).reset_index(drop=True)

# Paso 6: reportamos cuántas colonias fueron excluidas y cuál fue el valor del
# percentil 99 usado como umbral de referencia.
print(f'Colonias a excluir por criterios de casos extremos: {len(casos_extremos_excluidos):,}')
print(f'Umbral p99 de consumo total por hogar: {umbral_p99:.4f}')
casos_extremos_excluidos
```

```python
# Construimos el dataset semestral filtrado excluyendo los casos extremos identificados.
df_semestre_filtrado = df_semestre_stats.loc[~mask_casos_extremos].copy()

print(f'Filas antes del filtro: {len(df_semestre_stats):,}')
print(f'Filas excluidas: {mask_casos_extremos.sum():,}')
print(f'Filas después del filtro: {len(df_semestre_filtrado):,}')
df_semestre_filtrado.head()
```

## 9. Regresión lineal simple sobre la base limpia

En esta sección se implementa un **modelo de regresión lineal simple** para responder la pregunta de investigación: si el número de hogares por colonia permite predecir el consumo total semestral de agua.

La especificación estimada es:

`consumo_total_i = β0 + β1 * Sum_TotHog_i + ε_i`

donde:

- `β0` es el intercepto,
- `β1` es el coeficiente asociado al número de hogares,
- y `ε_i` representa el término de error.

El objetivo es reportar de manera explícita:

- **coeficientes estimados**,
- **coeficiente de determinación (R²)**,
- **métricas de error**,
- y una interpretación sustantiva del alcance del modelo frente a la problemática hídrica planteada.

Este modelo se ajusta sobre la **base filtrada**, después de excluir los casos que distorsionaban severamente la relación entre hogares y consumo.


```python
# Ajustamos una regresión lineal simple: consumo_total ~ hogares.
df_regresion = df_eda_limpio[['alcaldia', 'colonia', 'Sum_TotHog', 'consumo_total']].dropna().copy()

x = df_regresion['Sum_TotHog'].astype(float).to_numpy()
y = df_regresion['consumo_total'].astype(float).to_numpy()

pendiente, intercepto = np.polyfit(x, y, 1)
df_regresion['consumo_predicho'] = intercepto + pendiente * x
df_regresion['residual'] = df_regresion['consumo_total'] - df_regresion['consumo_predicho']

ss_res = np.sum((y - df_regresion['consumo_predicho']) ** 2)
ss_tot = np.sum((y - y.mean()) ** 2)
r2 = 1 - (ss_res / ss_tot)
rmse_reg = np.sqrt(np.mean(df_regresion['residual'] ** 2))
mae_reg = np.mean(np.abs(df_regresion['residual']))

coeficientes_regresion = pd.DataFrame({
    'termino': ['intercepto', 'hogares'],
    'coeficiente_estimado': [intercepto, pendiente]
})
coeficientes_regresion['coeficiente_estimado'] = coeficientes_regresion['coeficiente_estimado'].round(4)

metricas_regresion = pd.DataFrame({
    'metrica': ['R2', 'RMSE', 'MAE', 'n_observaciones'],
    'valor': [r2, rmse_reg, mae_reg, len(df_regresion)]
})
metricas_regresion['valor'] = metricas_regresion['valor'].astype(float).round(4)
metricas_regresion.loc[metricas_regresion['metrica'] == 'n_observaciones', 'valor'] = len(df_regresion)

print('Coeficientes estimados')
display(coeficientes_regresion)
print('Métricas de ajuste')
display(metricas_regresion)
```

### Interpretación del modelo de regresión lineal

El modelo lineal simple formaliza la relación entre el número de hogares y el consumo total semestral de agua a nivel colonia. El coeficiente estimado para `Sum_TotHog` es positivo, por lo que, en promedio, un mayor número de hogares se asocia con un mayor consumo agregado. Sin embargo, el valor de `R²` y las métricas de error muestran que el ajuste es sólo parcial. Esto implica que el número de hogares sí aporta señal explicativa, pero deja sin capturar una parte importante de la variación observada entre colonias. En relación con la problemática planteada, el resultado sugiere que esta variable funciona como un predictor base útil, aunque insuficiente para describir por completo la demanda de agua en una ciudad tan heterogénea como la CDMX. En otras palabras, la regresión sí sirve para el ejercicio, pero también hace visibles sus límites analíticos.


```python
# Visualizamos la recta ajustada sobre la nube de puntos limpia.
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df_regresion, x='Sum_TotHog', y='consumo_total', alpha=0.35, s=35)
plt.plot(
    np.sort(df_regresion['Sum_TotHog']),
    intercepto + pendiente * np.sort(df_regresion['Sum_TotHog']),
    color='crimson',
    linewidth=2,
    label='Recta de regresión'
)
plt.title('Regresión lineal: hogares vs consumo total semestral')
plt.xlabel('Hogares 2020')
plt.ylabel('Consumo total semestral')
plt.legend()
plt.tight_layout()
plt.show()
```

### Interpretación de la recta de regresión

La nube de puntos y la recta ajustada permiten visualizar de forma inmediata el alcance y los límites del modelo lineal simple. La pendiente positiva confirma que, en promedio, las colonias con más hogares concentran un mayor consumo total semestral, lo cual es coherente con la intuición demográfica básica. Sin embargo, la dispersión alrededor de la recta es amplia: para niveles similares de hogares aparecen colonias con consumos muy distintos. Esa separación entre puntos observados y línea estimada es importante porque muestra que la relación general existe, pero no es suficientemente cerrada como para hablar de una predicción precisa. Sustantivamente, la gráfica sugiere que el tamaño poblacional captura una parte del patrón agregado, pero que la demanda hídrica por colonia depende también de otras condiciones urbanas, económicas y de infraestructura que el modelo no está incorporando.

### 9.1 Regresión log-log

Como la relación entre hogares y consumo total puede no ser lineal en escala original, probamos una especificación log-log. Este modelo permite evaluar si el número de hogares captura mejor la variación relativa del consumo cuando ambas variables se expresan en logaritmos.

Además, la pendiente en este contexto se interpreta como una **elasticidad** aproximada: cuánto cambia porcentualmente el consumo ante un cambio porcentual en el número de hogares.


```python
# Ajustamos una regresión lineal en escala log-log.
df_regresion_log = df_eda_limpio[['alcaldia', 'colonia', 'Sum_TotHog', 'consumo_total']].dropna().copy()
df_regresion_log = df_regresion_log[(df_regresion_log['Sum_TotHog'] > 0) & (df_regresion_log['consumo_total'] > 0)].copy()

df_regresion_log['log_hogares'] = np.log(df_regresion_log['Sum_TotHog'].astype(float))
df_regresion_log['log_consumo_total'] = np.log(df_regresion_log['consumo_total'].astype(float))

x_log = df_regresion_log['log_hogares'].to_numpy()
y_log = df_regresion_log['log_consumo_total'].to_numpy()

pendiente_log, intercepto_log = np.polyfit(x_log, y_log, 1)
df_regresion_log['log_consumo_predicho'] = intercepto_log + pendiente_log * x_log
df_regresion_log['consumo_predicho_log_modelo'] = np.exp(df_regresion_log['log_consumo_predicho'])
df_regresion_log['residual_log'] = df_regresion_log['log_consumo_total'] - df_regresion_log['log_consumo_predicho']

ss_res_log = np.sum((y_log - df_regresion_log['log_consumo_predicho']) ** 2)
ss_tot_log = np.sum((y_log - y_log.mean()) ** 2)
r2_log = 1 - (ss_res_log / ss_tot_log)

resumen_regresion_log = pd.Series({
    'intercepto_log': intercepto_log,
    'pendiente_log_hogares': pendiente_log,
    'r2_log': r2_log,
    'n_observaciones': len(df_regresion_log)
}, name='valor').to_frame()

resumen_regresion_log
```

```python
# Visualizamos el ajuste log-log.
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df_regresion_log, x='log_hogares', y='log_consumo_total', alpha=0.35, s=35)
plt.plot(
    np.sort(df_regresion_log['log_hogares']),
    intercepto_log + pendiente_log * np.sort(df_regresion_log['log_hogares']),
    color='darkorange',
    linewidth=2,
    label='Recta log-log'
)
plt.title('Regresión log-log: log(hogares) vs log(consumo total)')
plt.xlabel('log(hogares)')
plt.ylabel('log(consumo total)')
plt.legend()
plt.tight_layout()
plt.show()
```

### Interpretación del modelo log-log

La especificación log-log permite leer la relación entre hogares y consumo como una asociación proporcional, no sólo absoluta. En esta escala, la pendiente resume una elasticidad aproximada: cuánto tiende a variar el consumo total cuando el número de hogares cambia en términos porcentuales. La utilidad de esta transformación es que comprime valores extremos y puede revelar una relación más estable si el crecimiento del consumo no es estrictamente lineal. Sin embargo, el resultado debe leerse junto con sus métricas y con la nube de puntos ajustada. Si el modelo log-log no mejora de manera clara el desempeño predictivo respecto al modelo lineal, la transformación funciona como exploración metodológica, no como solución central. En este caso, el modelo aporta una lectura complementaria útil y mantiene la conclusión de que el número de hogares explica sólo una fracción del comportamiento real del consumo.


### 9.2 Comparación básica de desempeño predictivo

Si la pregunta es si el número de hogares permite **predecir** el consumo total, no basta con inspeccionar el ajuste dentro de la muestra. Aquí comparamos un modelo lineal en escala original contra uno log-log usando una partición simple train/test.

El objetivo es medir si alguno de los dos modelos generaliza mejor cuando enfrenta datos no usados en el ajuste.


```python
# Funciones auxiliares para evaluación predictiva sin dependencias extra.
def train_test_split_simple(df, test_size=0.2, random_state=42):
    df = df.sample(frac=1, random_state=random_state).reset_index(drop=True)
    n_test = max(1, int(round(len(df) * test_size)))
    test = df.iloc[:n_test].copy()
    train = df.iloc[n_test:].copy()
    return train, test


def rmse(y_true, y_pred):
    y_true = np.asarray(y_true, dtype=float)
    y_pred = np.asarray(y_pred, dtype=float)
    return np.sqrt(np.mean((y_true - y_pred) ** 2))


def mae(y_true, y_pred):
    y_true = np.asarray(y_true, dtype=float)
    y_pred = np.asarray(y_pred, dtype=float)
    return np.mean(np.abs(y_true - y_pred))


def r2_score_manual(y_true, y_pred):
    y_true = np.asarray(y_true, dtype=float)
    y_pred = np.asarray(y_pred, dtype=float)
    ss_res = np.sum((y_true - y_pred) ** 2)
    ss_tot = np.sum((y_true - y_true.mean()) ** 2)
    return 1 - (ss_res / ss_tot)
```

```python
# Comparamos capacidad predictiva fuera de muestra: lineal vs log-log.
df_modelos = df_eda_limpio[['Sum_TotHog', 'consumo_total']].dropna().copy()
df_modelos = df_modelos[(df_modelos['Sum_TotHog'] > 0) & (df_modelos['consumo_total'] > 0)].copy()

train_df, test_df = train_test_split_simple(df_modelos, test_size=0.2, random_state=42)

# Modelo lineal en escala original
pendiente_train, intercepto_train = np.polyfit(train_df['Sum_TotHog'], train_df['consumo_total'], 1)
test_df = test_df.copy()
test_df['pred_lineal'] = intercepto_train + pendiente_train * test_df['Sum_TotHog']

# Modelo log-log
pendiente_train_log, intercepto_train_log = np.polyfit(np.log(train_df['Sum_TotHog']), np.log(train_df['consumo_total']), 1)
test_df['pred_loglog'] = np.exp(intercepto_train_log + pendiente_train_log * np.log(test_df['Sum_TotHog']))

comparacion_modelos = pd.DataFrame([
    {
        'modelo': 'lineal_original',
        'rmse': rmse(test_df['consumo_total'], test_df['pred_lineal']),
        'mae': mae(test_df['consumo_total'], test_df['pred_lineal']),
        'r2_test': r2_score_manual(test_df['consumo_total'], test_df['pred_lineal'])
    },
    {
        'modelo': 'log_log',
        'rmse': rmse(test_df['consumo_total'], test_df['pred_loglog']),
        'mae': mae(test_df['consumo_total'], test_df['pred_loglog']),
        'r2_test': r2_score_manual(test_df['consumo_total'], test_df['pred_loglog'])
    }
])

comparacion_modelos
```

```python
# Visualización observados vs predichos en test para ambos modelos.
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

sns.scatterplot(x=test_df['consumo_total'], y=test_df['pred_lineal'], alpha=0.5, s=35, ax=axes[0])
lims0 = [
    min(test_df['consumo_total'].min(), test_df['pred_lineal'].min()),
    max(test_df['consumo_total'].max(), test_df['pred_lineal'].max())
]
axes[0].plot(lims0, lims0, color='crimson', linestyle='--')
axes[0].set_title('Test: observados vs predichos (lineal)')
axes[0].set_xlabel('Consumo observado')
axes[0].set_ylabel('Consumo predicho')

sns.scatterplot(x=test_df['consumo_total'], y=test_df['pred_loglog'], alpha=0.5, s=35, ax=axes[1], color='darkorange')
lims1 = [
    min(test_df['consumo_total'].min(), test_df['pred_loglog'].min()),
    max(test_df['consumo_total'].max(), test_df['pred_loglog'].max())
]
axes[1].plot(lims1, lims1, color='crimson', linestyle='--')
axes[1].set_title('Test: observados vs predichos (log-log)')
axes[1].set_xlabel('Consumo observado')
axes[1].set_ylabel('Consumo predicho')

plt.tight_layout()
plt.show()
```

### Interpretación de observados vs predichos en test

Los gráficos de observados contra predichos permiten evaluar visualmente qué tan cerca quedan las predicciones de la línea ideal de 45 grados. Si un modelo predijera con alta precisión, la mayor parte de los puntos se concentraría muy cerca de esa diagonal. Aquí, en cambio, se observa una dispersión considerable alrededor de la línea, lo cual indica que ambos modelos cometen errores relevantes fuera de muestra. La comparación entre paneles sirve para detectar cuál especificación se aproxima mejor al patrón ideal: el modelo con puntos relativamente más cercanos a la diagonal y menor dispersión general es el que generaliza mejor. Aun así, la lectura sustantiva no cambia demasiado: incluso el modelo preferible muestra desajustes amplios, por lo que el número de hogares sí ayuda a anticipar el nivel de consumo, pero no basta para lograr predicciones territoriales finas sin incorporar variables adicionales.

### 9.3 Conclusión comparativa final

La comparación fuera de muestra entre el modelo lineal y el modelo log-log permite cerrar el análisis con un criterio más exigente que el simple ajuste visual. Si el modelo lineal obtiene menor error absoluto y menor error cuadrático medio en test, la lectura es que, para esta base y después del filtrado aplicado, la especificación original describe mejor el nivel agregado de consumo que la versión logarítmica. Si la diferencia entre ambos modelos es pequeña, eso también es informativo: indica que el número de hogares sí aporta una señal real, pero esa señal es limitada y no basta para producir predicciones finas por sí sola.

En términos sustantivos, el ejercicio muestra que `Sum_TotHog` funciona como un predictor base útil para aproximar la demanda total de agua por colonia, pero no como una explicación suficiente del fenómeno. La heterogeneidad territorial de la CDMX, la mezcla de usos del suelo, posibles fugas, diferencias socioeconómicas y patrones de consumo no observados siguen introduciendo una variación importante. Por eso, el hallazgo principal no es solamente cuál modelo “gana”, sino reconocer que ambos confirman la misma conclusión: el tamaño de la colonia importa, pero está lejos de agotar la estructura real del consumo de agua.


## 10. Mapa de desempeño predictivo por colonia

Para traducir el error del modelo a una lectura territorial, construimos un mapa temático por colonia con tres clases:

- **Funciona (±10%)**: la predicción cae dentro de un margen de error absoluto de 10% respecto al consumo observado.
- **Subestima el gasto**: el modelo predice más de 10% por debajo del consumo observado.
- **Sobreestima el gasto**: el modelo predice más de 10% por encima del consumo observado.

Para el mapa usamos el modelo con mejor desempeño fuera de muestra según la comparación anterior. Las **manzanas** se muestran como referencia espacial tenue y las **colonias** se colorean según la categoría de error.


```python
# Mapa por colonia: funciona / subestima / sobreestima con margen razonable de ±10%.
from matplotlib.patches import Patch

manzanas_path = base_path / 'poligono_manzanas_cdmx' / 'poligono_manzanas_cdmx.shp'

gdf_colonias_mapa = gpd.read_file(hogares_path)[['cve_col', 'alcaldia', 'colonia', 'geometry']].copy()
gdf_manzanas = gpd.read_file(manzanas_path)

# Elegimos el modelo que mejor generaliza fuera de muestra (menor MAE; si empata, menor RMSE).
modelo_mapa = comparacion_modelos.sort_values(['mae', 'rmse']).iloc[0]['modelo']

# Consolidamos a nivel colonia real (cve_col), porque el dataset analítico todavía contiene
# algunos nombres duplicados que apuntan a la misma colonia cartográfica.
df_mapa = df_eda_limpio[['alcaldia', 'colonia', 'cve_col', 'Sum_TotHog', 'consumo_total']].dropna().copy()
df_mapa = df_mapa[(df_mapa['Sum_TotHog'] > 0) & (df_mapa['consumo_total'] > 0)].copy()
df_mapa = (
    df_mapa.groupby('cve_col', as_index=False)
    .agg(
        alcaldia=('alcaldia', 'first'),
        colonia=('colonia', 'first'),
        Sum_TotHog=('Sum_TotHog', 'first'),
        consumo_total=('consumo_total', 'sum')
    )
)

if modelo_mapa == 'log_log':
    pendiente_mapa, intercepto_mapa = np.polyfit(np.log(df_mapa['Sum_TotHog']), np.log(df_mapa['consumo_total']), 1)
    df_mapa['consumo_predicho'] = np.exp(intercepto_mapa + pendiente_mapa * np.log(df_mapa['Sum_TotHog']))
else:
    pendiente_mapa, intercepto_mapa = np.polyfit(df_mapa['Sum_TotHog'], df_mapa['consumo_total'], 1)
    df_mapa['consumo_predicho'] = intercepto_mapa + pendiente_mapa * df_mapa['Sum_TotHog']

margen_razonable = 0.10
df_mapa['error_pct'] = (df_mapa['consumo_predicho'] - df_mapa['consumo_total']) / df_mapa['consumo_total']

df_mapa['categoria_prediccion'] = np.select(
    [
        df_mapa['error_pct'].abs() <= margen_razonable,
        df_mapa['error_pct'] < -margen_razonable,
        df_mapa['error_pct'] > margen_razonable,
    ],
    [
        'Funciona (±10%)',
        'Subestima el gasto',
        'Sobreestima el gasto',
    ],
    default='Sin clasificar'
)

# Llevamos todo a una proyección métrica para mostrar escalas en kilómetros.
crs_metrico = 'EPSG:32614'
gdf_colonias_mapa = gdf_colonias_mapa.to_crs(crs_metrico)
gdf_manzanas = gdf_manzanas.to_crs(crs_metrico)

gdf_mapa = gdf_colonias_mapa.merge(
    df_mapa[['cve_col', 'consumo_total', 'consumo_predicho', 'error_pct', 'categoria_prediccion']],
    on='cve_col',
    how='left'
)
gdf_mapa['en_modelo'] = gdf_mapa['categoria_prediccion'].notna()
gdf_mapa['categoria_prediccion'] = gdf_mapa['categoria_prediccion'].fillna('Fuera del análisis / entrenamiento')

resumen_mapa = (
    gdf_mapa.loc[gdf_mapa['en_modelo'], 'categoria_prediccion']
    .value_counts()
    .rename_axis('categoria')
    .reset_index(name='n_colonias')
)

n_fuera = (~gdf_mapa['en_modelo']).sum()

display(pd.DataFrame({
    'modelo_seleccionado_para_mapa': [modelo_mapa],
    'margen_razonable_pct': [margen_razonable * 100],
    'n_colonias_en_modelo': [int(gdf_mapa['en_modelo'].sum())],
    'n_colonias_fuera_del_analisis': [int(n_fuera)]
}))
display(resumen_mapa)

palette_mapa = {
    'Funciona (±10%)': '#2E8B57',
    'Subestima el gasto': '#E67E22',
    'Sobreestima el gasto': '#8E44AD',
    'Fuera del análisis / entrenamiento': '#D9D9D9',
}
orden_categorias = [
    'Funciona (±10%)',
    'Subestima el gasto',
    'Sobreestima el gasto',
    'Fuera del análisis / entrenamiento'
]

fig, ax = plt.subplots(figsize=(12, 12))

# Base espacial: manzanas como referencia visual tenue.
gdf_manzanas.boundary.plot(ax=ax, linewidth=0.08, color='lightgray', alpha=0.25)

for categoria in orden_categorias:
    subset = gdf_mapa[gdf_mapa['categoria_prediccion'] == categoria]
    if not subset.empty:
        subset.plot(
            ax=ax,
            color=palette_mapa[categoria],
            edgecolor='white',
            linewidth=0.20 if categoria != 'Fuera del análisis / entrenamiento' else 0.10,
            alpha=0.95 if categoria != 'Fuera del análisis / entrenamiento' else 0.75,
            label=categoria
        )

legend_handles = [Patch(facecolor=palette_mapa[c], edgecolor='white', label=c) for c in orden_categorias]
ax.legend(handles=legend_handles, title='Lectura de la predicción', loc='lower left', frameon=True)
ax.set_title('CDMX: desempeño predictivo del gasto de agua por colonia', fontsize=14)
ax.text(
    0.02,
    0.02,
    (
        f'Margen razonable: ±{margen_razonable:.0%} | Modelo usado: {modelo_mapa}\n'
        'Gris claro: colonias no utilizadas en el análisis o entrenamiento final'
    ),
    transform=ax.transAxes,
    fontsize=10,
    bbox=dict(facecolor='white', alpha=0.85, edgecolor='none')
)

# Escalas en kilómetros para que la figura no quede sin referencia espacial.
xmin, xmax = ax.get_xlim()
ymin, ymax = ax.get_ylim()
xticks = np.linspace(xmin, xmax, 5)
yticks = np.linspace(ymin, ymax, 5)
ax.set_xticks(xticks)
ax.set_yticks(yticks)
ax.set_xticklabels([f'{x/1000:.0f}' for x in xticks])
ax.set_yticklabels([f'{y/1000:.0f}' for y in yticks])
ax.set_xlabel('Coordenada Este (km, UTM 14N)')
ax.set_ylabel('Coordenada Norte (km, UTM 14N)')
ax.grid(color='gray', linestyle=':', linewidth=0.4, alpha=0.35)

plt.tight_layout()
plt.show()
```

### Interpretación del mapa de desempeño predictivo

El mapa territorializa el error del modelo y permite observar que el desempeño predictivo no se distribuye de manera homogénea en la ciudad. Las colonias coloreadas en verde representan casos en los que la predicción cae dentro del margen razonable de ±10%, mientras que los tonos de subestimación y sobreestimación muestran zonas donde el modelo se desvía con mayor claridad. Esta lectura ayuda a identificar que el error no es sólo un problema estadístico agregado, sino también espacial: hay territorios donde la relación entre hogares y consumo se comporta de manera distinta al promedio. Las colonias en gris claro corresponden a unidades no utilizadas en el análisis o entrenamiento final, ya sea por exclusión metodológica o por no formar parte de la base depurada. En conjunto, el mapa refuerza la conclusión de que el número de hogares funciona como aproximación inicial, pero no como descriptor suficiente del patrón hídrico capitalino.
